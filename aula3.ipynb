## Erro Absoluto

```python
import math

valor_exato = math.pi
valor_aproximado = 3.14

erro_absoluto = abs(valor_exato - valor_aproximado)

print(f'Valor exato      : {valor_exato}')
print(f'Valor aproximado : {valor_aproximado}')
print(f'Erro absoluto    : {erro_absoluto}')
```

## Erro Relativo

```python
erro_relativo = erro_absoluto / abs(valor_exato)

print(f'Erro relativo: {erro_relativo}')
print(f'Erro relativo (%): {erro_relativo*100:.6f}%')
```

## Exemplo

```python
x = 2112.9
x_aprox = 2113

y = 5.2
y_aprox = 5.3

EAx = abs(x - x_aprox)
EAy = abs(y - y_aprox)

ERx = EAx / abs(x)
ERy = EAy / abs(y)

print('Erro absoluto de x:', EAx)
print('Erro relativo de x:', ERx)
print('Erro absoluto de y:', EAy)
print('Erro relativo de y:', ERy)
```

## Truncamento

$$
\operatorname{trunc}(x,n)=
\frac{\left\lfloor 10^n \cdot x \right\rfloor}{10^n}
$$

A função piso, representada por $\lfloor x \rfloor$, retorna o **maior número inteiro menor ou igual a** $x$.

Na fórmula do truncamento, primeiro multiplicamos o número por $10^n$, aplicamos a função piso para remover a parte fracionária e, em seguida, dividimos novamente por $10^n$ para restaurar a escala original.

```python
import math

numero = 2.7182818

math.trunc(numero)
```

```python
def truncar(numero, n=2):
    fator = 10 ** n
    return math.trunc(numero * fator) / fator


print('Número original:', numero)
print('Truncado:', truncar(numero,5))
```

## Arredondamento para o par (IEEE 754)

```python
numero = 2.7182818

print('Número original:', numero)
print('Arredondado:', round(numero,4))
```

```python
n = 4

print("Arredondamento para baixo")
print(f"2.71824 = {round(2.71824, n)}")

print("\nArredondamento para cima")
print(f"2.71826 = {round(2.71826, n)}")

print("\nEmpate (último dígito mantido é par)")
print(f"2.71825 = {round(2.71825, n)}")

print("\nEmpate (último dígito mantido é ímpar)")
print(f"2.71835 = {round(2.71835, n)}")
```

## Comparando os erros

```python
valor = 2.7182818

trunc = truncar(valor,4)
arred = round(valor,4)

EA_trunc = abs(valor-trunc)
EA_arred = abs(valor-arred)

print('Erro truncamento :', EA_trunc)
print('Erro arredondamento:', EA_arred)
```

## Comparando números em ponto flutuante

Em vez de comparar dois números utilizando `==`, é comum verificar se a diferença entre eles é menor que uma tolerância previamente definida.

Essa ideia será utilizada nos métodos numéricos para decidir quando uma aproximação é suficientemente boa.

```python
tolerancia = 1e-10

resultado = 0.1 + 0.2

print(f"Resultado: {resultado}")
print(f"Erro: {abs(resultado - 0.3)}")


if abs(resultado - 0.3) < tolerancia:
    print("Os valores são considerados iguais.")
else:
    print("Os valores são diferentes.")
```
