# Zeros Reais de Funções Reais

Um zero (ou raiz) de uma função é um valor ξ tal que

**f(ξ)=0**

## Exemplo

Considere

**f(x)=x²−4**

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return x**2 - 4

x = np.linspace(-4, 4, 400)

plt.plot(x, f(x))
plt.axhline(0, color="black")
plt.grid(True)
plt.show()
```

Neste caso,

x²−4=0

Logo,

ξ₁=-2

ξ₂=2


Nem sempre conseguimos determinar as raízes exatamente.

Nesses casos, utilizizamos métodos numéricos para obter aproximações.

# Localização/Isolamento das Raízes

Objetivo:

Encontrar um intervalo [a,b] que contenha pelo menos uma raiz.

Considere

**f(x)=x²−4**

Calcule:

f(1)

f(3)


```python
def f(x):
    return x**2 - 4

print(f(1))
print(f(3))
```

Como

f(1)·f(3)<0

existe pelo menos uma raiz no intervalo [1,3].

## Exemplo

Considere

**f(x)=x³−9x+3**


```python
def f(x):
    return x**3 - 9*x + 3

x = np.linspace(-4, 4, 400)

plt.plot(x, f(x))
plt.axhline(0, color="black")
plt.grid(True)
plt.show()
```

Construa uma tabela de valores para localizar as raízes.

```python
def f(x):
    return x**3 - 9*x + 3

def df(x):
    return 3*x**2 - 9

xs = [-100,-10,-5,-3,-1,0,1,2,3,5,10,100]

print(f"{'x':>6}{'f(x)':>12}{"f'(x)":>12}")
for x in xs:
    print(f"{x:>6}{f(x):>12}{df(x):>12}")
```

Mudanças de sinal:

[-5,-3]

[0,1]

[2,3]

Sinal da derivada:

positivo

negativo

positivo


Intervalo I₁=[-5,-3] contém ξ₁

Intervalo I₂=[0,1] contém ξ₂

Intervalo I₃=[2,3] contém ξ₃

Cada intervalo contém uma única raiz.


```python
def f(x):
    return x**3 - 9*x + 3

xs = [-5, -3, -1, 0, 1, 3, 5]
ys = [f(x) for x in xs]

plt.figure(figsize=(8,4))
plt.plot(xs, ys, 'o-')
plt.axhline(0, color='black')
plt.grid(True)
plt.show()
```

```python
def f(x):
    return x**3 - 9*x + 3

x = np.linspace(-5, 5, 1000)
y = f(x)

plt.figure(figsize=(8,4))
plt.plot(x, y)
plt.axhline(0, color='black')
plt.grid(True)
plt.show()
```
