# Fatoração LU

## Ideia

Decompomos a matriz de coeficientes em duas matrizes triangulares:

$$
A=LU,\qquad
L=\begin{bmatrix}1&0&0\\m_{21}&1&0\\m_{31}&m_{32}&1\end{bmatrix},\qquad
U=\begin{bmatrix}u_{11}&u_{12}&u_{13}\\0&u_{22}&u_{23}\\0&0&u_{33}\end{bmatrix}
$$

Para resolver $Ax=b$, calculamos primeiro $Ly=b$ (substituição progressiva) e depois $Ux=y$ (substituição retroativa). Se $A$ não mudar, reutilizamos $L$ e $U$ para outros vetores $b$.

Nesta implementação, usamos **eliminação sem pivotamento parcial**, supondo pivôs não nulos.

## Algoritmo: Cálculo de L e U

```text
Entrada:
    A = matriz quadrada n × n

U = A
L = I (identidade n × n)

Para k de 1 até n−1 faça
    Para i de k+1 até n faça
        m = U[i,k] / U[k,k]
        L[i,k] = m
        U[i,k] = 0
        Para j de k+1 até n faça
            U[i,j] = U[i,j] − m * U[k,j]
        Fim Para
    Fim Para
Fim Para

Retorne L, U
```

````python
def fatoracao_lu(A):
    n = len(A)
    U = [linha[:] for linha in A]
    L = [[1 if i == j else 0 for j in range(n)] for i in range(n)]

    for k in range(n - 1):
        if U[k][k] == 0:
            raise ValueError("Pivô nulo: seria necessário pivotamento.")
        for i in range(k + 1, n):
            m = U[i][k] / U[k][k]
            L[i][k] = m
            U[i][k] = 0
            for j in range(k + 1, n):
                U[i][j] -= m * U[k][j]

    return L, U
````

## Algoritmo: Resolução de Ax = b

```text
Entrada:
    L, U = matrizes triangulares n × n
    b = vetor de dimensão n

# Substituição progressiva (Ly = b)
Para i de 1 até n faça
    s = 0
    Para j de 1 até i−1 faça
        s = s + L[i,j] * y[j]
    Fim Para
    y[i] = b[i] − s
Fim Para

# Substituição retroativa (Ux = y)
Para i de n até 1 faça
    s = 0
    Para j de i+1 até n faça
        s = s + U[i,j] * x[j]
    Fim Para
    x[i] = (y[i] − s) / U[i,i]
Fim Para

Retorne x
```

````python
def resolver_lu(L, U, b):
    n = len(b)
    y = [0] * n
    for i in range(n):
        s = 0
        for j in range(i):
            s += L[i][j] * y[j]
        y[i] = b[i] - s  # diagonal de L igual a 1

    x = [0] * n
    for i in range(n - 1, -1, -1):
        s = 0
        for j in range(i + 1, n):
            s += U[i][j] * x[j]
        x[i] = (y[i] - s) / U[i][i]

    return y, x
````

## Exemplo 1

$$
\begin{cases}
3x_1+2x_2+4x_3=1\\
x_1+x_2+2x_3=2\\
4x_1+3x_2+2x_3=3
\end{cases}
$$

$$
A=\begin{bmatrix}3&2&4\\1&1&2\\4&3&2\end{bmatrix},\qquad
b=\begin{bmatrix}1\\2\\3\end{bmatrix}
$$

````python
A = [
    [3, 2, 4],
    [1, 1, 2],
    [4, 3, 2]
]
b = [1, 2, 3]

L, U = fatoracao_lu(A)
print("L:")
for linha in L:
    print(linha)
print("U:")
for linha in U:
    print(linha)

y, x = resolver_lu(L, U, b)
print("y:", y)
print("x:", x)
````

### Mesmo A, novo vetor b

$$
b_{\mathrm{novo}}=\begin{bmatrix}9\\5\\9\end{bmatrix}
$$

Reutilizamos os mesmos fatores $L$ e $U$.

````python
b_novo = [9, 5, 9]
y_novo, x_novo = resolver_lu(L, U, b_novo)
print("Novo y:", y_novo)
print("Novo x:", x_novo)
````

## Exemplo 2

Resolva o sistema usando fatoração LU **sem pivotamento parcial**:

$$
\begin{cases}
3x_1-4x_2+x_3=9\\
x_1+2x_2+2x_3=3\\
4x_1-3x_3=-2
\end{cases}
$$

Em seguida, reutilize $L$ e $U$ para o novo vetor:

$$
b_{\mathrm{novo}}=\begin{bmatrix}3\\5\\1\end{bmatrix}
$$

````python
A_ex = [
    [3, -4, 1],
    [1, 2, 2],
    [4, 0, -3]
]
b_ex = [9, 3, -2]

L_ex, U_ex = fatoracao_lu(A_ex)
print("L:")
for linha in L_ex:
    print(linha)
print("U:")
for linha in U_ex:
    print(linha)

y_ex, x_ex = resolver_lu(L_ex, U_ex, b_ex)
print("y:", y_ex)
print("x:", x_ex)
````

````python
b_ex_novo = [3, 5, 1]
y_ex_novo, x_ex_novo = resolver_lu(L_ex, U_ex, b_ex_novo)
print("Novo y:", y_ex_novo)
print("Novo x:", x_ex_novo)
````
