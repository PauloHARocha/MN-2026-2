# Método da Posição Falsa

## Ideia do método

Desejamos encontrar uma aproximação $\bar{x}$ para a raiz $\xi$ de uma função

$$
f(x)=0
$$

Sabemos que

$$
\xi \in [a,b]
$$

e que

$$
f(a)\cdot f(b)<0.
$$

A cada iteração:

- calculamos a interseção da reta secante com o eixo $x$

$$
x=\frac{af(b)-bf(a)}{f(b)-f(a)}
$$

- escolhemos o subintervalo que contém a raiz;
- repetimos o processo até atingir a precisão desejada.


## Algoritmo

```text
Entrada:
    intervalo inicial [a,b]
    precisões ε₁ e ε₂
    número máximo de iterações K

k = 0

enquanto (k < K)

    x = (a·f(b) − b·f(a)) / (f(b) − f(a))

    se |f(x)| < ε₂
        retorne x

    se f(a) · f(x) < 0
        b = x
    senão
        a = x

    se (b − a) < ε₁
        retorne (a + b) / 2

    k = k + 1

retorne (a + b) / 2
```

````python
import math

def posicao_falsa(f, a, b, eps1=1e-3, eps2=1e-3, kmax=100):
    k = 0
    while k < kmax:
        x = (a * f(b) - b * f(a)) / (f(b) - f(a))
        if abs(f(x)) < eps2:
            return x, k + 1
        if f(a) * f(x) < 0:
            b = x
        else:
            a = x
        if (b - a) < eps1:
            return (a + b) / 2, k + 1
        k += 1
    return (a + b) / 2, k
````

## Exemplo 1

$
f(x)=x^2-2,\qquad [a,b]=[1,2],\qquad \varepsilon_1=\varepsilon_2=0,1.
$

````python
f=lambda x:x**2-2
posicao_falsa(f,1,2,eps1=0.1,eps2=0.1)
````

## Exemplo 2

$
f(x)=x^3-9x+3,\qquad [a,b]=[0,1],\qquad \varepsilon_1=\varepsilon_2=5\\times10^{-4}.
$

````python
f=lambda x:x**3-9*x+3
posicao_falsa(f,0,1,eps1=5e-4,eps2=5e-4)
````

## Exercício

$
f(x)=x^3-x-2,\qquad [a,b]=[1,2],\qquad \varepsilon_1=\varepsilon_2=0,2.
$

````python
f=lambda x:x**3-x-2
posicao_falsa(f,1,2,eps1=0.2,eps2=0.2)
````

## Análise da Convergência

A seguir registramos o histórico das iterações para construir tabelas e visualizar a convergência do método.

````python
def posicao_falsa_historico(f,a,b,eps1=1e-3,eps2=1e-3,kmax=100):
    ks=[];A=[];B=[];X=[]
    k=1
    while k<=kmax:
        ks.append(k);A.append(a);B.append(b)
        x=(a*f(b)-b*f(a))/(f(b)-f(a));X.append(x)
        if abs(f(x))<eps2: break
        if f(a)*f(x)<0: b=x
        else: a=x
        if (b-a)<eps1: break
        k+=1
    return ks,A,B,X
````

````python
import pandas as pd

k,a,b,x=posicao_falsa_historico(f,1,2,eps1=0.2,eps2=0.2)
df=pd.DataFrame({"k":k,"a":a,"b":b,"x":x})
df["a'"]=df["a"].shift(-1)
df["b'"]=df["b"].shift(-1)
df["L"]=df["b'"]-df["a'"]
df.head()
````

````python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
plt.plot(df["k"],df["L"],marker="o")
plt.axhline(0.2,linestyle="--")
plt.xlabel("Iteração")
plt.ylabel("Comprimento do intervalo")
plt.title("Convergência do Método da Posição Falsa")
plt.grid(True)
plt.show()
````
