# Aula 10 - Comparação entre Métodos

Comparação entre Bisseção, Posição Falsa, Ponto Fixo, Newton-Raphson e Secante.

```python
import numpy as np
import pandas as pd
```

## Implementações

```python
def bissecao(f,a,b,eps=1e-6,kmax=100):
    hist=[]
    for k in range(1,kmax+1):
        x=(a+b)/2
        hist.append((k,x,f(x)))
        if abs(f(x))<eps or abs(b-a)<eps: return x,hist
        if f(a)*f(x)<0: b=x
        else: a=x
    return x,hist

def falsa_posicao(f,a,b,eps=1e-6,kmax=100):
    hist=[]
    for k in range(1,kmax+1):
        x=(a*f(b)-b*f(a))/(f(b)-f(a))
        hist.append((k,x,f(x)))
        if abs(f(x))<eps: return x,hist
        if f(a)*f(x)<0: b=x
        else: a=x
    return x,hist

def ponto_fixo(phi,f,x,eps=1e-6,kmax=100):
    hist=[]
    for k in range(1,kmax+1):
        xn=phi(x)
        hist.append((k,xn,f(xn)))
        if abs(f(xn))<eps or abs(xn-x)<eps: return xn,hist
        x=xn
    return x,hist

def newton(f,df,x,eps=1e-6,kmax=100):
    hist=[]
    for k in range(1,kmax+1):
        xn=x-f(x)/df(x)
        hist.append((k,xn,f(xn)))
        if abs(f(xn))<eps or abs(xn-x)<eps: return xn,hist
        x=xn
    return x,hist

def secante(f,x0,x1,eps=1e-6,kmax=100):
    hist=[]
    for k in range(1,kmax+1):
        x2=x1-f(x1)*(x1-x0)/(f(x1)-f(x0))
        hist.append((k,x2,f(x2)))
        if abs(f(x2))<eps or abs(x2-x1)<eps: return x2,hist
        x0,x1=x1,x2
    return x2,hist
```

## Exemplo 19

```python
f=lambda x:x**3-x-1
df=lambda x:3*x**2-1
phi=lambda x:(x+1)**(1/3)

metodos={}
metodos["Bisseção"]=bissecao(f,1,2,1e-6)[1]
metodos["Posição Falsa"]=falsa_posicao(f,1,2,1e-6)[1]
metodos["Ponto Fixo"]=ponto_fixo(phi,f,1,1e-6)[1]
metodos["Newton"]=newton(f,df,0,1e-6)[1]
metodos["Secante"]=secante(f,0,0.5,1e-6)[1]
rows=[]
for nome,h in metodos.items():
    k,x,fx=h[-1]
    rows.append([nome,x,fx,len(h)])
pd.DataFrame(rows,columns=["Método","x","f(x)","Iterações"])
```

## Exemplo 18



$f(x)=e^{-x^2}-\cos(x)$

$f'(x)=-2xe^{-x^2}+\sin(x)$

$\phi(x)=\cos(x)-e^{-x^2}+x$


```python
f=lambda x: np.exp(-x**2)-np.cos(x)
df=lambda x:-2*x*np.exp(-x**2)+np.sin(x)
phi=lambda x: np.cos(x)-np.exp(-x**2)+x

eps = 1e-4

valores_iniciais = {
    "intervalo": (1, 2),
    "x0": 1.5,
    "newton": 1.5,
    "secante": (1, 2)
}

metodos = {
    "Bisseção": bissecao(
        f,
        valores_iniciais["intervalo"][0],
        valores_iniciais["intervalo"][1],
        eps
    )[1],

    "Posição Falsa": falsa_posicao(
        f,
        valores_iniciais["intervalo"][0],
        valores_iniciais["intervalo"][1],
        eps
    )[1],

    "Ponto Fixo": ponto_fixo(
        phi,
        f,
        valores_iniciais["x0"],
        eps
    )[1],

    "Newton": newton(
        f,
        df,
        valores_iniciais["newton"],
        eps
    )[1],

    "Secante": secante(
        f,
        valores_iniciais["secante"][0],
        valores_iniciais["secante"][1],
        eps
    )[1]
}

rows = []

for nome, historico in metodos.items():
    k, x, fx = historico[-1]
    rows.append([nome, x, fx, len(historico)])

pd.DataFrame(
    rows,
    columns=["Método", "x", "f(x)", "Iterações"]
)
```
