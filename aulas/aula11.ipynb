# Eliminação de Gauss

## Ideia

Desejamos transformar um sistema linear

$
Ax=b
$

em um sistema triangular superior equivalente, que poderá ser resolvido por substituição retroativa.


## Algoritmo

```text
Entrada:
    matriz A
    vetor b

Para k = 0 até n−2

    Para i = k+1 até n−1

        m = A[i][k] / A[k][k]

        Para j = k até n−1
            A[i][j] = A[i][j] − m·A[k][j]

        b[i] = b[i] − m·b[k]

Resolver o sistema triangular superior.
```


````python
def sistema_triangular(A, b):
    n = len(A)

    x = [0] * n

    x[n-1] = b[n-1] / A[n-1][n-1]

    for k in range(n-2, -1, -1):
        s = 0
        for j in range(k + 1, n):
            s += A[k][j] * x[j]

        x[k] = (b[k] - s) / A[k][k]

    return x
````

## Algoritmo

```text
Entrada:
    A = matriz triangular superior n × n
    b = vetor de termos constantes

Para k de 1 até n−1 faça

    Para i de k+1 até n faça

        m = A[i,k] / A[k,k]
        A[i,k] = 0

        Para j de k+1 até n faça

            A[i,j] = A[i,j] − m * A[k,j]
        Fim Para

        b[i] = b[i] − m * b[k]

    Fim Para
Fim Para
Retorne A, b
```


````python
def gauss(A, b):
    n = len(A)

    A = [linha[:] for linha in A]
    b = b[:]

    for k in range(n - 1):
        for i in range(k + 1, n):
            m = A[i][k] / A[k][k]

            for j in range(k, n):
                A[i][j] -= m * A[k][j]

            b[i] -= m * b[k]

    return A, b
````

## Exemplo 1

````python
A = [
    [3, 2, 4],
    [1, 1, 2],
    [4, 3, -2]
]

b = [1, 2, 3]
````

````python
A_tri, b_tri = gauss(A, b)

print("Matriz triangular:")
for linha in A_tri:
    print(linha)

print("\nVetor b:")
print(b_tri)
````

````python
x = sistema_triangular(A_tri, b_tri)

print("Solução:")
print(x)
````
