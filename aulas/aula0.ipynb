## Como um computador calcula π?

As bibliotecas Python possuem constantes matemáticas prontas:

- `math.pi`
- `numpy.pi`

Mas como essas bibliotecas obtêm esse valor?

Uma possibilidade é utilizar a série de Gregory-Leibniz para aproximar π.

```python
import math
import numpy as np

print(f"math.pi:  {math.pi}")
print(f"numpy.pi: {np.pi}")
```

A série de Gregory-Leibniz é dada por

$$
\pi = 4\left(1-\frac13+\frac15-\frac17+\cdots\right)
$$

Quanto mais termos utilizarmos, melhor será a aproximação.

```python
n = 1_000_000

pi_aproximado = 0

for k in range(n):
    pi_aproximado += 4 * (-1)**k / (2 * k + 1)

print(f"Gregory-Leibniz: {pi_aproximado}")
```

Veremos diversos algoritmos para aproximar soluções de problemas matemáticos, como:

- encontrar raízes de funções;
- resolver sistemas de equações lineares;
- interpolar dados;
- calcular integrais numericamente.

A ideia central é sempre a mesma: utilizar algoritmos para obter aproximações suficientemente precisas de forma eficiente.
