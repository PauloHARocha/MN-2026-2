## Conversão entre bases

```python
numero = 347

print(f"Decimal: {numero}")
print(f"Binário: {bin(numero)}")
print(f"Octal:   {oct(numero)}")
print(f"Hex:     {hex(numero)}")
```

## Binário para decimal

```python
binario = "1010"

decimal = int(binario, 2)

print(f"{binario}₂ = {decimal}₁₀")
```

Experimente outros valores:

```python
print(int("0F", 2))
print(int("AB", 16))
print(int("11111111", 2))
```

## Decimal para binário

```python
numeros = [13, 42, 100, 347]

for n in numeros:
    print(f"{n}₁₀ = {bin(n)[2:]}₂")
```

A função `bin()` retorna uma string iniciada por `0b`.

```python
print(bin(42))
print(bin(42)[2:])
```

## Representação de números fracionários

```python
print(0.5)
print(0.75)
print(0.125)

print()
print("Em Python utiliza-se ponto como separador decimal.")
```
